## 調査の中断・再開

ここまでが今回の試作です。同じランタイムなら、準備セルを再実行しても保存済みの正解を引き継げます。
Colabのランタイムが削除されると保存ファイルも消えるため、残したい進捗をダウンロードしてください。
再開時は準備セルを実行し、次の「進捗を読み込む」からファイルを選びます。


```python
# @title 進捗をダウンロードする
from google.colab import files
if lab.lesson.progress_path.exists():
    files.download(str(lab.lesson.progress_path))
else:
    print('まだ保存された回答がありません。')
```

```python
# @title 進捗を読み込む（再開時のみ）
import json
from google.colab import files
uploaded = files.upload()
if len(uploaded) == 1:
    lab.lesson.restore(json.loads(next(iter(uploaded.values()))))
    print('進捗を復元しました。正解済み:', len(lab.lesson.answers))
else:
    print('進捗ファイルを1つ選んでください。')
```
